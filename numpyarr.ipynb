```python
import numpy as np 
myarr = np.array([[1,2,3], [4,5,6], [6,7,8], [9, 10, 11]]) 
flat = myarr.flatten()
flat
```

```python
# indexing : 
print(flat[0])
print(flat[-5])
```

```python
# slicing :
print(flat[:3], flat[::4], flat[1:5])
```

```python
import numpy as np
import time
size = 1_000
l1 = list(range(size))
l2 = list(range(size)) 
start = time.time()
mul = [x*y for x, y in zip(l1,l2)]
end = time.time()
print("time taken : ", end-start)
print(mul[1:10])
```

```python
import numpy as np
import time
size = 1_000
n1 = np.array(list(range(size)))
n2 = np.array(list(range(size))) 
start = time.time()
mul_np = n1 * n2 
end = time.time()
print("time taken : ", end-start)
print(mul_np[1:10]) 
```
